```python
import pandas as pd
from pprint import pprint
import os
import re
from tqdm import tqdm
import json
import requests
import matplotlib.pyplot as plt
from collections import defaultdict
import spacy

nlp = spacy.load('de_core_news_lg') # oder de_core_news_sm


def flatten(xss):
    return [x for xs in xss for x in xs]

plots_dir = "plots/"
try:
    os.mkdir(plots_dir)
    print(f"Directory '{plots_dir}' created successfully.")
except FileExistsError:
    print(f"Directory '{plots_dir}' already exists.")
```

```python
with open("generated_data/Leipzig_KI_sentences_all.json", "r") as f:
    leipzig_data = json.load(f)
taz24_data = pd.read_csv("generated_data/KI_sentences_translated.csv")
print("leipzig:", len(leipzig_data),list(leipzig_data.keys()), "\ntaz24:", taz24_data.columns)
```

```python
# build regerex for anthropomorphic verbs
verb_synonyms = list(cognitive_v_lexicon.keys()) + [cognitive_v_lexicon[k]["synonyms"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["synonyms"]) > 0]
verb_synonyms = [','.join(syn) if type(syn) == list else syn for syn in verb_synonyms]
verb_synonyms = ','.join(verb_synonyms).split(",")
verb_synonyms = [x.lower() for x in verb_synonyms]
anthro_regex = "".join(["\\b"+x.replace(" ","\\s")+"\\b|" for x in verb_synonyms])[:-1]
print(anthro_regex)
```

```python
anthro_taz = {}
for i,sent in enumerate(taz24_data["original"]):       
    if len(sent) > 0 and re.search(anthro_regex, sent):
        #print("res:",sent, re.search(anthro_regex, sent).string)
        anthro_taz[taz24_data["id"][i]] = sent

print(len(anthro_taz))
print(anthro_taz)

with open(r'generated_data/taz_anthro_re.json', 'w') as f:
    json.dump(anthro_taz, f)
```

```python
# search with stemm/lemma
def get_lemma(sent):
    doc = nlp(sent)
    return [token.lemma for token in doc]
    
```

```python
verb_lemmas = [get_lemma(x) for x in verb_synonyms]
verb_lemmas = flatten(verb_lemmas)
```

```python
anthro_taz_l = {}
for i,sent in enumerate(taz24_data["original"]):
    for l in get_lemma(sent):
        if len(sent) > 0 and l in verb_lemmas:
            anthro_taz_l[taz24_data["id"][i]] = sent
print(len(anthro_taz_l))

with open(r'generated_data/taz_anthro_l_re.json', 'w') as f:
    json.dump(anthro_taz, f)
```

```python
anthro_leipzig = {}
for k in leipzig_data.keys():
    sents = []
    for l_id in leipzig_data[k].keys():
        sent = leipzig_data[k][l_id]
        if re.search(anthro_regex, sent):
            #print("res:",sent, re.search(anthro_regex, sent).string)
            sents.append({l_id:sent})
        anthro_leipzig[k]=sents

print(len(anthro_leipzig))

with open(r'generated_data/leipzig_anthro_re.json', 'w') as f:
    json.dump(anthro_leipzig, f)
```

```python
anthro_leipzig_l = {}
for k in leipzig_data.keys():
    sents = []
    for l_id in leipzig_data[k].keys():
        sent = leipzig_data[k][l_id]
        for l in get_lemma(sent):
            if l in verb_lemmas:
                if l_id not in [list(x.keys())[0] for x in sents]:
                    sents.append({l_id:sent})
        anthro_leipzig_l[k]=sents
    anthro_leipzig_l

print(len(anthro_leipzig_l))
print(anthro_leipzig_l['2003_1M'])


with open(r'generated_data/leipzig_anthro_l_re.json', 'w') as f:
    json.dump(anthro_leipzig_l, f)
```

```python
# Get all sentences by year
all_sentences = []
for year in anthro_leipzig:
    all_sentences.extend(anthro_leipzig[year])

# Get all sentences with lemmas by year
all_sentences_l = []
for year in anthro_leipzig_l:
    all_sentences_l.extend(anthro_leipzig_l[year])
```

```python
# combine results and save
def combine_results(a_taz, a_leipzig, filename):
    all_anthro = {}
    for key, value in a_taz.items():
        all_anthro[key] = value

    for k in a_leipzig.keys():
        for elem in a_leipzig[k]:
            for key, value in elem.items():
                all_anthro[key] = value

    with open(f'generated_data/{filename}.json', 'w') as f:
        json.dump(all_anthro, f)
```

```python
combine_results(anthro_taz, anthro_leipzig,'anthro_re')
combine_results(anthro_taz_l, anthro_leipzig_l,'anthro_re_l')
```

```python
# group by verbs found

# Count verbs in taz24 anthropomorphic sentences
verb_counts = defaultdict(int)

for sent_id, sent in anthro_taz.items():
    sent_lower = sent.lower()
    for verb in verb_synonyms:
        verb_pattern = r'\b' + verb.replace(" ", r'\s') + r'\b'
        if re.search(verb_pattern, sent_lower):
            verb_counts[verb] += 1

# Sort by frequency
sorted_verbs = sorted(verb_counts.items(), key=lambda x: x[1], reverse=True)

# Create dictionary with percentages
taz24_verbs_grouped = {}
for verb, count in sorted_verbs:
    percentage = count / len(anthro_taz)
    taz24_verbs_grouped[verb] = {
        "count": count,
        "percentage": round(percentage, 2)
    }

# Display
pprint(taz24_verbs_grouped)
```
