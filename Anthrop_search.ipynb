## Regex of Anthropomorphisms 

Based on anthropomorphic markers. First experiment: verbs from previous literature, and their synonyms (based on wiktionary)

- Anthroscore:
    - why verbs: E.2 Parts of Speech. Moreover, we find that 55% and 44% of S↑ and S↓ respectively are those in which the masked entity is the subject of the verb (rather than the object). In S↓, when the masked entity is the subject, it is often with intransitive verbs, which are less likely to suggest that the masked entity is exhibiting behavioral potential and directly acting upon another entity (the object of the sentence)
    - cognitive verbs: know, think, believe, understand, remember, forget, guess, pretend, dream, mean, suspect, suppose, feel, assume
    - top verbs for A > 1: achieve, learn, guide, show, embed, fool, find, need, assist, follow, search, mislead, inspire, win, demonstrate, benefit, try, face, deceive, plan, make, steer, generative, attempt, retrain, train, flow, weight, require, alternate focus, motivate, experiment, tackle, see, hide, spiking, recommend, discover, participate, spike, pass, code, check, suggest, decide, interference, aim, move 
    - Using the weighted log-odds ratio method described in Section 5.1, we compute whether the differences in frequency for these words are statistically significant (z-score >1.96, which corresponds to a 95% CI.) We find that among these verbs, only understand is statistically significantly more frequent in S↑

```python
# build regerex for anthropomorphic verbs
verb_synonyms = list(cognitive_v_lexicon.keys()) + [cognitive_v_lexicon[k]["synonyms"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["synonyms"]) > 0]
#verb_synonyms = verb_synonyms + [cognitive_v_lexicon[k]["antonyms"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["antonyms"]) > 0]
#verb_synonyms = verb_synonyms + [cognitive_v_lexicon[k]["sinnverwandt"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["sinnverwandt"]) > 0]
#verb_synonyms = verb_synonyms + [cognitive_v_lexicon[k]["hypernyms"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["hypernyms"]) > 0]
#verb_synonyms = verb_synonyms + [cognitive_v_lexicon[k]["hyponyms"] for k in cognitive_v_lexicon.keys() if len(cognitive_v_lexicon[k]["hyponyms"]) > 0]
verb_synonyms = [','.join(syn) if type(syn) == list else syn for syn in verb_synonyms]
verb_synonyms = ','.join(verb_synonyms).split(",")
verb_synonyms = [x.lower() for x in verb_synonyms]
anthro_regex = "".join(["\\b"+x.replace(" ","\\s")+"\\b|" for x in verb_synonyms])[:-1]
print(anthro_regex)
```

```python
# writes a matcher with DependencyMatcher to match verbs + sub or verbs + obj
# currently not used
def write_matcher():
    matcher = DependencyMatcher(nlp.vocab)
    
    # Pattern for verb with subject (nsubj or nsubjpass)
    verb_subject_pattern = [
        {
            "RIGHT_ID": "verb",
            "RIGHT_ATTRS": {"POS": "VERB"}
        },
        {
            "LEFT_ID": "verb",
            "REL_OP": ">",
            "RIGHT_ID": "subject",
            "RIGHT_ATTRS": {"DEP": {"IN": ["nsubj", "nsubjpass"]}}
        }
    ]
    
    # Pattern for verb with direct object (obj or iobj)
    verb_object_pattern = [
        {
            "RIGHT_ID": "verb",
            "RIGHT_ATTRS": {"POS": "VERB"}
        },
        {
            "LEFT_ID": "verb",
            "REL_OP": ">",
            "RIGHT_ID": "object",
            "RIGHT_ATTRS": {"DEP": {"IN": ["obj", "iobj"]}}
        }
    ]
    
    matcher.add("verb_subject", [verb_subject_pattern])
    matcher.add("verb_object", [verb_object_pattern])
    print('matcher', len(matcher))
    return matcher
    
# Find verb + obj/subj
def find_verb_relations(doc):
    relations_sb = []
    relations_ob = []
    if doc.has_annotation("DEP"):
        for token in doc:
            if token.pos_ == "VERB" and token.lemma in verb_lemmas:
                for child in token.children:
                    if child.dep_ in ["sb"]:
                        relations_sb.append({
                            "type": "verb_subject",
                            "verb": token.text,
                            "dependent": child.text,
                            "dep_label": child.dep_,
                            "verb_pos": token.pos_,
                            "dep_pos": child.pos_,
                            "sentence": doc.text
                        })
                    # Akkusativobjekt (I, II), Dativ, Präpositionalobjekte, Genitivobjekt, klausales Objekt
                    elif child.dep_ in ["oa", "oa2", "da", "op", "og", "oc"]:
                        relations_ob.append({
                            "type": "verb_object",
                            "verb": token.text,
                            "dependent": child.text,
                            "dep_label": child.dep_,
                            "verb_pos": token.pos_,
                            "dep_pos": child.pos_,
                            "sentence": doc.text
                        })
    
    return relations_sb, relations_ob


#extra Fall: vokativ, Anrede wie in "Hans, sag ...", für AI-terms filtern
def find_vokativ(doc):
    res = []
    if doc.has_annotation("DEP"):
        for token in doc:
            if token.pos_ == "VERB" and token.lemma in verb_lemmas:
                for child in token.children:
                    if child.dep_ in ["vo"]:
                        res.append({
                            "type": "verb_subject",
                            "verb": token.text,
                            "dependent": child.text,
                            "dep_label": child.dep_,
                            "verb_pos": token.pos_,
                            "dep_pos": child.pos_,
                            "sentence": doc.text
                        })
    return res
```

```python
# filter sentences that are in a verb + subj or verb + obj structure
anthro_taz_dep_sb = {}
anthro_taz_dep_ob = {}
anthro_taz_dep_v = {}
for sent_id, sent in anthro_taz_l.items():
    doc = nlp(sent)
    # obj/subj
    relations_sb, relations_ob = find_verb_relations(doc)
    if relations_sb:
        anthro_taz_dep_sb[sent_id] = relations_sb
    if relations_ob:
        anthro_taz_dep_ob[sent_id] = relations_ob
    # vokativ
    relations_v = find_vokativ(doc)
    if relations_v:
        anthro_taz_dep_v[sent_id] = relations_v

print(len(anthro_taz_dep_sb))
print(len(anthro_taz_dep_ob))
print(len(anthro_taz_dep_v))

anthro_taz_dep = {'sb': anthro_taz_dep_sb, 'ob': anthro_taz_dep_ob, 'v': anthro_taz_dep_v}

with open(r'generated_data/anthro_taz_dep.json', 'w') as f:
    json.dump(anthro_taz_dep, f)
```

```python
# combine results and save
def combine_results(a_taz, a_leipzig, filename):
    all_anthro = {}
    try:
        for key, value in a_taz.items():
            if key.endswith('txt'):
                all_anthro[key] = value
    except:
        print(f"{filename} a_leipzig is not a dict")
        all_anthro.update(a_leipzig)
    try:
        for elem in a_leipzig:
            for key, value in elem.items():
                all_anthro[key] = value
    except:
        print(f"{filename} a_leipzig is not a dict")
        all_anthro.update(a_leipzig)

    with open(f'generated_data/{filename}.json', 'w') as f:
        json.dump(all_anthro, f)
        
    return all_anthro
```
